```python
#W4A10:Longest Collatz up to N:

def collatz(n):
    count=1
    if n == 1:
        return count
    while n > 1:
        if n % 2 == 0:
            n /= 2
            count +=1
        else:
            n = n * 3 + 1
            count +=1
    return count
n=int(input())
max_collatz=0
min_number=0
for i in range(1,n+1):
    if collatz(i) > max_collatz or (collatz(i)==max_collatz and i < min_number ):
        min_number=i
        max_collatz=collatz(i)
print(min_number, max_collatz)
```

[LongestCollatzUpToN]:
Nhập vào số nguyên dương n. Tìm số x trong [1, n] có độ dài dãy Collatz lớn nhất và in ra x cùng độ dài dãy tương ứng (x L). Nếu có nhiều số cùng độ dài, chọn số nhỏ nhất."

```python
def collatz_length(x):
    count = 1
    while x != 1:
        if x % 2 == 0:
            x //= 2
        else:
            x = 3 * x + 1
        count += 1
    return count

def find_longest_collatz(n):
    max_length = 0
    result = 1
    for i in range(1, n + 1):
        length = collatz_length(i)
        if length > max_length or (length == max_length and i < result):
            max_length = length
            result = i
    return result, max_length

n = int(input("Nhập số nguyên dương n: "))
x, L = find_longest_collatz(n)
print(f"({x}, {L})")
```

ReverseAddPalindrome:
Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số đối xứng (palindrome). In ra số bước và giá trị palindrome thu được

```python
#Hàm kiểm tra xem có phải là số đối xứng chưa
def is_palindrome(n):
    return str(n) == str(n)[::-1]
#Hàm đảo ngược số
def reverse_number(n):
    return int(str(n)[::-1])
#Hàm đếm số bước nếu chưa phải là số đối xứng
def find_palindrome(n):
    steps = 0
    while not is_palindrome(n):
        n += reverse_number(n)
        steps += 1
    return steps, n

# Nhập từ người dùng
n = int(input("Nhập số nguyên dương n: "))
steps, result = find_palindrome(n)
print(f"{steps} {result}")
```
