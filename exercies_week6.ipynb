W6A10:
Viết chương trình nhập vào một danh sách các số nguyên, và số nguyên dương k. In ra list gồm các cặp số khác nhau có tổng bằng k. Mỗi cặp là một tuple, hai số trong tuple có
thể giống nhau, thứ tự xuất hiện trong danh sách của số thứ nhất nhỏ hơn thứ tự của số thứ hai. Các tuple trong list được sắp xếp theo thứ tự tăng dần của phần tử đầu tiên trong tuple.

```python
def find_unique_pairs_with_sum_k():
    nums = list(map(int, input("Nhập danh sách số nguyên: ").split()))
    k = int(input("Nhập số nguyên dương k: "))

    result = []
    seen = set()
    n = len(nums)

    for i in range(n):
        for j in range(i + 1, n):
            if nums[i] + nums[j] == k:
                pair = (nums[i], nums[j])
                if pair not in seen:
                    result.append(pair)
                    seen.add(pair)

    result.sort(key=lambda x: x[0])
    print(result)

# Gọi hàm
find_unique_pairs_with_sum_k()
```

W6A12:[Tìm phần tử xuất hiện nhiều nhất]
Nhập vào một list các số nguyên. Hãy tìm số xuất hiện nhiều lần nhất trong list. Nếu có nhiều số cùng có số lần xuất hiện nhiều nhất, hãy in ra số nhỏ nhất trong số đó.

```python
lst=list(map(int, input().split()))
d={}
for i in lst:
    if i not in d:
        d[i]=1
    else:
        d[i] +=1
max_num=0
count= 0
for key, value in d.items():
    if value > max_num or ( value == max_num and key < count):
        max_num = value 
        count = key
print(count)
```

W6A15:[Lọc dictionary theo giá trị]
Nhập vào một dictionary có key là chuỗi và value là số nguyên, và một số nguyên k ở dòng tiếp theo. Hãy tạo một dictionary mới chỉ chứa các cặp keyvalue từ dictionary ban đầu mà có value lớn hơn k. In ra dictionary mới.

```python
n=list(input().split())
k=int(input())
d={}
for i in range(0,len(n),2):
    if int(n[i+1]) > k:
        d[n[i]]= int(n[i+1])
print(d)
```

Nhập vào mảng vuông 2 chiều n hàng, n cột, in ra đường chéo chính và đường chéo phụ của mảng

```python
n=int(input())
lst=[]
for i in range(n):
    row=list(map(int, input().split()))
    lst.append(row)
cheo_chinh=[]
cheo_phu=[]
for i in range(n):
    cheo_chinh.append(lst[i][i])
    cheo_phu.append(lst[i][n-i-1])
print("Đường chéo chính là:", *cheo_chinh)
print("Đường chéo phụ là:" , *cheo_phu)
```
