W8A1:[Cylinder]
Xây dựng lớp Cylinder (hình trụ) gồm: bán kính và chiều cao (kiểu số thực) với hai phương thức: tính diện tích bề mặt (diện tích toàn phần) và thể tích của hình trụ
Đầu vào:
- Đâu vào từ bàn phím gồm một dòng duy nhất chứa bán kính và chiều cao của hình trụ, cách nhau bởi một dấu cách.
Đầu ra
- In ra màn hình diện tích bề mặt và thể tích của hình trụ (sử dụng lại những hàm vừa viết, π=3.14). (Kết quả làm tròn đến 2 chữ số sau dấu phẩy).

```python
class Cylinder:
    def __init__(self, radius, height):
        self.radius = radius
        self.height = height
        self.pi = 3.14
    def dttp (self):
        return 2 * self.pi * self.radius * (self.radius + self.height)
    def volume (self):
        return self.pi * (self.radius ** 2) * self.height

r, h = map(int, input().split())
c = Cylinder(r, h)
print(f"{c.dttp():.2f}, {c.volume():.2f}")
```

W8A2:[Date]
Xây dựng lớp Date biểu diễn ngày tháng năm với hai phương thức kiểm tra xem một ngày tháng nào đó có hợp lệ hay không và phương thức tính ngày kế tiếp của ngày tháng năm nào đó nếu nó là hợp lệ
Nhập vào một ngày tháng năm, (ngày tháng năm biểu diễn dưới dạng dd/mm/yyyy).
Nếu ngày tháng năm không tồn tại, in ra dòng chữ INVALID.
Nếu là hợp lệ in ra ngày kế tiếp của nó.
- Đầu vào:
Đầu vào từ bàn phim chứa dữ liệu ngày, tháng, năm dưới dạng dd/mm/yyyy.
- Đầu ra:
In ra màn hình ngày kế tiếp của ngày vừa nhập dưới dạng dd/mm/yyyy. Nếu ngày tháng không tồn tại, in ra INVALID

```python
def nhuan(year):
    if year % 4 == 0:
        if year % 400 == 0:
            return True
        elif year % 100 == 0: 
            return False
        return True
n_31 = [1, 3, 5, 7, 8, 10, 12]
n_30 = [4, 6, 9, 11]
class Date:
    def __init__ (self, date):
        self.day = int(date[:2])
        self.month = int(date[3:5])
        self.year = int(date[6:])    
    def valid (self):
        if self.month > 12: return False
        elif self.month == 2:
            if nhuan(self.year):
                if self.day == 29: return True
                else: return self.day == 28
        elif self.month in n31:
            return self.day == 31
        elif self.month in n30:
            return self.day == 30    
    def ngayketiep(self):
        if self.day == 31 and self.month == 12:
            s = str(self.year + 1)
            while len(s) < 4:
                s = "0" + s
            return f"01/01/{s}"
        elif (self.month in n31 and self.day == 31) or (self.month in n30 and self.day == 30): 
            if self.month < 9: return f"01/0{self.month + 1}/{self.year}"
            return f"01/{self.month + 1}/{self.year}"
        if self.day < 9:
            return f"0{self.day + 1}/{self.month}/{self.year}"
        else: return f"{self.day + 1}/{self.month}/{self.year}"
s = input()
day = Date(s)
if day.valid == False:
    print("INVALID")
else:
    print(day.ngayketiep())
```

Xây dựng lớp ShoppingCart (giỏ hàng) để quản lý các sản phẩm của khách. Dùng danh sách để lưu các sản phẩm, mỗi sản phẩm là một từ điển có name (tên) và price(giá).
Với các phương thức sau:
1. Khởi tạo với danh sách rỗng
2. Thêm vào giỏ hàng một sản phẩm mới
3. Xóa khỏi giỏ một sản phẩm nào đó (tìm theo tên để xóa)
4. Kiểm tra xem giỏ hàng có rỗng không
5. Tính tổng tiền giỏ hàng
6. Hiển thị toàn bộ sản phẩm trong giỏ
7. Xóa toàn bộ giỏ hàng
Tạo menu cho người dùng nhập lựa chọn truy vấn từ 2 tới 7. Chương trình dừng khi người dùng nhập "yes" cho câu hỏi "Có muốn thoát khỏi chương trình không?"

```python
class ShoppingCart:
    def __init__ (self):
        self.cart = {}
        self.total = 0
    def Add (self, product, price):
        self.cart[product] = price
        self.total += price
    def Remove(self, product):
        self.total -= self.cart[product]
        del self.cart[product]
    def Check_empty(self):
        return len(self.cart) == 0
    def Total_money(self):
        return self.total
    def Hienthi(self):
        for product, price in self.cart.items():
            print(f"{product}: {price}")
    def Clear(self):
        self.cart.clear()
cart = ShoppingCart()
while True:
    q = int(input())
    if q == 2:
        product, price = map(str, input().split())
        cart.Add(product, int(price))
    elif q == 3:
        product = input()
        cart.Remove(product)
    elif q == 4:
        if cart.Check_empty(): print("Rỗng")
        else: print("Không rỗng")
    elif q == 5:
        print(cart.Total_money())
    elif q == 6:
        cart.Hienthi()
    elif q == 7:
        cart.Clear()   
    print("Bạn có muốn dừng không?")
    s = input()
    if s == 'yes':
        break
```

W8A7:
- Yêu cầu:
Lớp Fraction(num, den) (tử, mẫu nguyên, mẫu ≠ 0) với:
Rút gọn phân số trong __init__ (quy ước mẫu > 0).
Các phương thức add/sub/mul/div trả về phân số đã rút gọn. __str__() in dạng a/b.
Input:
- a b op c d (tương ứng a/b (op) c/d, op thuộc {+,-,*,/})
Output:
- Phân số kết quả đã rút gọn

```python
from math import gcd, lcm
def xuli(tu, mau):
    ok = 0
    if tu * mau < 0: ok = 1
    tu = abs(tu)
    mau = abs(mau)
    k = gcd(tu, mau)
    tu = tu // k
    mau = mau // k
    if ok:
        return (-tu, mau)
    return (tu, mau)
class Fraction: 
    def __init__ (self, tu, mau):
        self.tu, self.mau = xuli(tu, mau)

    def add(self, other):
        mauchung = lcm(self.mau, other.mau)
        tu = self.tu * (mauchung // self.mau) + other.tu * (mauchung // other.mau)
        self.tu, self.mau = xuli(tu, mauchung)

    def sub(self, other):
        mauchung = lcm(self.mau, other.mau)
        tu = self.tu * (mauchung // self.mau) - other.tu * (mauchung // other.mau)
        self.tu, self.mau = xuli(tu, mauchung)
    
    def mul(self, other):
        self.tu, self.mau = xuli(self.tu * other.tu, self.mau * other.mau)

    def div(self, other):
        self.tu, self.mau = xuli(self.tu * other.mau, self.mau * other.tu)

    def __str__(self):
        return f"{self.tu}/{self.mau}"

s = list(map(str, input().split()))
a = int(s[0])
b = int(s[1])

p1 = Fraction(a, b)
c = int(s[3])
d = int(s[4])

p2 = Fraction(c, d)
if s[2] == '+':
    p1.add(p2)
elif s[2] == '-':
    p1.sub(p2)
elif s[2] == '*':
    p1.mul(p2)
else:
    p1.div(p2)

print(p1)
```

```python
class BankAccount:
    def __init__ (self, owner, balance):
        self.owner = owner
        self.balance = balance
    def deposit (self, money):
        self.balance += money
    def withdraw(self, money):
        if self.balance >= money:
            self.balance -= money
    def __str__(self):
        return f"{self.balance:.2f}"
name, ib = map(str, input().split())
ib = float(ib)
owner1 = BankAccount(name, ib)
for _ in range(int(input())):
    s1, mon = map(str, input().split())
    mon = float(mon)
    if s1 == "DEPOSIT":
        owner1.deposit(mon)
    elif s1 == "WITHDRAW":
        owner1.withdraw(mon)
print(owner1)
```

-Yêu cầu:
Lớp Student(name) với: add_score(subject, score)
avg() trả về điểm trung bình. rank() trả về "Excellent" nếu avg ≥ 8, "Good" nếu 6.5–<8, "Average" nếu 5–<6.5, ngược lại "Poor".
Input:
- Dòng 1: name
- Dòng 2: n
- n dòng tiếp: subject score (score là số thực 0–10)
Output:
- name avg rank (avg 2 chữ số).

```python
class Student:
    def __init__ (self, name):
        self.name = name
        self.total = 0
        self.cnt = 0
    def add_score(self, subject, score):
        self.total += score
        self.cnt += 1
    def avg(self):
        return f"{self.total / self.cnt:.2f}"
    def rank(self):
        t = self.total / self.cnt
        if t >= 8: return "Excellent"
        elif t < 8 and t >= 6.5: return "Good"
        elif t >= 5 and t < 6.5: return "Average"
        else: return "Poor"
name = input()
s1 = Student(name)
for _ in range(int(input())):
    s, sc = map(str, input().split())
    sc = float(sc)
    s1.add_score(s, sc)
print(f"{name} {s1.avg()} {s1.rank()}")
```

Góc cần in ra được ký hiệu là PHI, và phải dùng độ (degrees), không dùng radian.
Dạng Input:

Dạng Output:
- In ra góc giữa hai mặt phẳng, chính xác tới hai chữ số thập phân.

```python
import math
class toado:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z
    def vec(self, other):
        x = other.x - self.x
        y = other.y - self.y
        z = other.z - self.z
        ans = toado(x, y, z)
        return ans
    def pt(self, other):
        x = self.y * other.z - self.z * other.y
        y = self.z * other.x - self.x * other.z
        z = self.x * other.y - self.y * other.x
        ans = toado(x, y, z)
        return ans
    def length(self):
        return math.sqrt(self.x ** 2 + self.y ** 2 + self.z ** 2)
    def __str__(self):
        return f"{self.x} {self.y} {self.z}"
x, y, z = map(float, input().split())
a = toado(x, y, z)
x, y, z = map(float, input().split())
b = toado(x, y, z)
x, y, z = map(float, input().split())
c = toado(x, y, z)
x, y, z = map(float, input().split())
d = toado(x, y, z)
v1 = a.vec(b)
v2 = a.vec(c)
v3 = b.vec(c)
v4 = b.vec(d)
v5 = v1.pt(v2)
v6 = v3.pt(v4)
t = (v5.x * v6.x + v5.y * v6.y + v5.z * v6.z) / (v5.length() * v6.length())
t = math.acos(t)
t = math.degrees(t)
print(f"{t:.2f}")
```

Viết hàm khởi tạo ComplexNumber(double initRealPart, double initImaginaryPart) để khởi tạo các giá trị phần thực và phần ảo của một số phức và hàm ComplexNumber multiply(complexNumber other) trả về tích của số phức hiện tại và một số phức khác.
Đầu vào:
-Đầu vào từ bàn phím gồm hai dòng, mỗi dòng gồm hai số thực cách nhau bởi một dấu cách biểu diễn phần thực và phần ảo của một số phức.

Đầu ra:
- In ra màn hình hai số biểu diễn phần thực và phần ảo của tích hai số phức.


```python
class ComplexNumber:
    def __init__(self, realPart, imaginaryPart):
        self.realPart = realPart
        self.imaginaryPart = imaginaryPart
    def ComplexNumbermultiply(self, other):
        return [self.realPart * other.realPart - self.imaginaryPart * other.imaginaryPart, self.realPart * other.imaginaryPart + self.imaginaryPart * other.realPart]   
x1, x2 = map(int, input().split())
x3, x4 = map(int, input().split())
t1 = ComplexNumber(x1, x2)
t2 = ComplexNumber(x3, x4)
print(*t1.ComplexNumbermultiply(t2))
```

W8A13: [Triangle]
Khai báo cấu trúc dữ liệu tam giác \(Triangle\) chứa \(3\) biến \(egde1,\,edge2,\,edge3\) kiểu \ (int\) là độ dài \(3\) cạnh của tam giác đó.
Viết hàm khởi tạo Triangle(int initEdge1,int initEdge2, int initEdge3) để khởi tạo độ dài các cạnh của tam giác và hàm double getArea() trả về diện tích của tam giác đó (nếu tam giác không hợp lệ, trả về \(-1\)).
Hoàn thiện chương trình dưới đây nhập vào \(3\) cạnh của một tam giác và in ra diện tích của tam giác đó.
Đầu vào:
- Một dòng duy nhất từ bàn phím chứa \(3\) số nguyên là độ dài \(3\) cạnh của tam giác
Đầu ra:
- In ra màn hình diện tích tam giác (làm tròn đến\ (2\) chữ số thập phân sau dấu phẩy). In ra \ (invalid\) nếu \(3\) số nhập vào không phải là \(3\) cạnh của một tam giác

```python
from math import sqrt
def check(a, b, c):
    return (a + b > c) & (a + c > b) & (b + c > a)
def dt(a, b, c):
    p = (a + b + c) / 2
    return math.sqrt(p * (p - a) * (p - b) * (p - c))
class Triangle:
    def __init__(self, edge1, edge2, edge3):
        self.edge1 = edge1
        self.edge2 = edge2
        self.edge3 = edge3
    def getArea(self):
        if not check(self.edge1, self.edge2, self.edge3):
            return -1
        return f"{dt(self.edge1, self.edge2, self.edge3):.2f}"   
a, b, c = map(int, input().split())
tri = Triangle(a, b, c)
if tri.getArea() == -1: print("invalid")
else: print(tri.getArea())
```
